# 9章ドキュメンテーション

ドキュメンテーションはDSで見落とされがちで、プロジェクトの終わりまで放置され、次の作業に移ると軽視・省略されやすい。しかしコードを再現可能にする重要な要素であり、他人に使ってもらうにも、将来自分が戻ってくるにも、良いドキュメントが欠かせない。最初の考えをすべて覚えておくのは不可能なので、記録を残す必要がある。
良いドキュメントは考えをうまく伝えるもので、まず「誰のために書くのか」を考えることが重要だ。プロジェクトを引き継ぐデータサイエンティスト向けか、チームの他のメンバー向けか、半年後の自分向けか。想定読者に合わせて詳細レベルや言葉を選ぶ。
優れたドキュメントの条件は次のとおり。

- 最新であること — コードの変更と同時に更新する。ログインが必要な独自ツールは避け、更新しやすくしておく。
- 構造化されていること — 最も重要な情報を冒頭に置くなど、すぐ見つかるようにする。

良いドキュメントは時間を大きく節約し、プロジェクトの複雑さを減らす。既存の作業を把握できれば重複を避けられ、新しいプロジェクトにも素早く取りかかれ、1年前の作業も思い出しやすい。
なお、以上はすべてのプロジェクトに共通だが、DS特有の留意点もある。DSでは複数の解法を試して最良のものに落ち着くのが一般的なので、実験や意思決定に至る思考プロセスを記録しておくとよい。後の質問対応や、将来の再検討に役立つ。

次の問に答えられる文章であるとよい

- このプロジェクトで使用したデータをなぜ選んだのか
- データについてどのような仮定を置いているのか
- なぜ他の分析方法ではなく、この分析方法を選んだのか
- この分析方法が機能しない状況はあるのか
- 改善できるようなショートカット（近道）はあるだろうか
- 今後、このプロジェクトに携わる人に提案したい、実験の道筋はほかにあるか
- このプロジェクトから学んだことは何か
