# 7章 テスト

## 7.1 テストを書くべき理由

テスト駆動開発

テスト駆動開発（TDD：Test-DrivenDevelopment）とは、開発者がプロジェクトの目的を達成する「オブジェクトや関数等のコードを**書く前に**テストを作成しておく」というソフトウェア開発プロセスである。その目的は、テストを使ってコードの要件を記述することであり、期待される振る舞いと関数の入出力をテストに指定する。そして、開発者はテストに合格するようにコードを書く。これにより、開発者はコードをモジュール化し、複雑な問題をテストしやすい断片に分解することができる。  
このアプローチはデータサイエンス（DS）のプロジェクトではうまくいかない場合が多い。なぜなら、一般的なDSプロジェクトの開始時点では、どのような関数が必要なのか正確にはわからないからだ。探索的な作業から始めることが多く、コードがどのように構造化されるべきかがわかるのはプロジェクトの終盤になってからだ。TDDについてもっと知りたいのであれば、Michal Jaworski＋TarekZiadé著『エキスパートPythonプログラミング改訂3版』（KADOKAWA）がTDDの観点からPython開発をカバーしている。

## 7.3 テストの書き方と実行方法

### 7.3.1 基本的なテスト

_pytest_ のドキュメント（ https://oreil.ly/slpmH ）に記載されているように、テストは4つの段階に分けて構成する。

1. アレンジ：例えば、データをロードするなど、関数を実行するのに必要なものをすべてセットする
2. 実行：テストを実行する
3. アサート：関数の実行結果が期待通りであることをチェックする
4. クリーンアップ：テストが痕跡を残さないようにする。→ ファイルを開いていた場合は必ず閉じるなど

### 7.3.2 予期しない入力のテスト

役に立ちそうなライブラリ
- Faker（ https://oreil.ly/A7Wb- ）：タイムスタンプやメールアドレスなど、様々な形式のフェイクデータを作成する  
    → テストコードに表示すべきでない個人データや機密データを扱う場合に重要
- Hypothesis（ https://oreil.ly/uTNW0 ）：コードが機能しないエッジケースの発見に役立つ  
    → プロパティベースのテストを実行し、さまざまなシナリオを記述したテストを書くことが出来る

## 7.5 テストの種類

**単体(ユニット)テスト**（unit test）  
関数やクラスといったコードの単位が、開発者の期待どおりに動くかどうかをチェックする

**統合(インテグレーション)テスト** (integration test)  
より大きなシステムが正しく接続されていることを確認する

**受け入れテスト** (acceptance test)  
システムがユーザーの期待どおりに動くかどうかをチェックする

**負荷テスト** (load test)  
データ量がユーザー数を増えても、システムは正常に機能することを確認する

**セキュリティテスト** (security test)  
システムが攻撃に対して耐性があるかどうかをチェックする

**ユーザビリティテスト** (usability test)  
システムが直感的に使えるかどうかをチェックする

## 7.6 データ検証

### 7.6.1 データ検証の例

一般的なデータ検証の手順は次の通りである。
- データセットに必要な列がすべて存在することをテスト
- 各列のデータが正しい型であることをテスト
- カラムがすべてnull値でないこと、あるいはnull値が一定の割合以上ないことをテスト
- データの統計的特性が期待どおりであることをチェック（データの平均、標準偏差、範囲など）
- 前項と関連するが、データ分布が前日、前週、といった関連する期間のデータと類似していることをチェック
- 異常なデータをチェック

### 7.6.2 Panderaを使ったデータ検証

_Pandera_ モジュールは軽量なデータ検証のツールで、スキーマを使ってpandasのDataFrameなどのデータ構造（Dask DataFrameなど類似のものも含む）を検証する。

データを検証するために、DataFrameがどのようにカラムを含むべきかを、各カラムのデータ型とともに定義するスキーマを定義する。
```python
import pandera.pandas as pa

schema = pa.DataFrameSchema({
    "Year": pa.Column(int),
    "India": pa.Column(float),
    "United States of America": pa.Column(float)
})
```

### 7.6.3 Pydanticによるデータ検証

```python
from pydantic import BaseModel
from typing import Optional
from datetime import datetime


class CountryData(BaseModel):
    country_name: str
    population: int
    literacy_rate_2020: Optional[float]
    timestamp: Optional[datetime] = None
```

- country_name: 必須かつ、文字列もしくはエラーを出さずに文字列にキャストできるものでなければならない
- population: 必須かつ、整数
- literacy_rate_2020: 任意かつ、浮動小数点数もしくはエラーを出さずに浮動小数点数にキャストできるものでなければならない
- timestamp: 任意かつ、datetimeオブジェクトもしくはdatetimeオブジェクトにキャストできるものでなければならない。デフォルト値はNone

````python
from pydantic import BaseModel
from typing import Optional
from datetime import datetime


class CountryData(BaseModel):
    country_name: str
    population: int
    literacy_rate_2020: Optional[float]
    timestamp: Optional[datetime] = None
````

````python
sample_data_correct = {
    'country_name': 'India',
    'population': 1417242151,
    'literacy_rate_2020': 79.43,
    'timestamp': datetime.now()
}

india = CountryData(**sample_data_correct)
````
