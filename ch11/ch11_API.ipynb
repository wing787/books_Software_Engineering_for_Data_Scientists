### 11.1.1 HTTPメソッドとステータスコード

一般的なAPIエンドポイント  
| エンドポイント（メソッド） | 機能 |
| --- | --- |
| GET | データを取得 |
| POST | データを送信して何かを生成、レスポンスを受信 |
| PUT | すでに存在しているものを更新するためにデータを送信 |
| DELETE | すでに存在するものを削除 |

一般的なHTTPステータスコード
| ステータスコード | 意味 |
| --- | --- |
| 2xx | リクエスト成功（例. 200 OK） |
| 4xx | クライアント側のエラーが発生した。例. エンドポイントのパスのスペルミス。→ レスポンスとして404エラー（404 Not Found）が返される |
| 5xx | サーバーにエラーが発生した。例えば、APIへのリクエスト時にAPIコードにバグが発生した |

他のHTTPリクエストはWikipediaの[List of HTTP status codes](https://ja.wikipedia.org/wiki/HTTP%E3%82%B9%E3%83%86%E3%83%BC%E3%82%BF%E3%82%B9%E3%82%B3%E3%83%BC%E3%83%89)を参照すること。

#### ヘッダーとURLパラメータ

- ヘッダーは、サーバーへのHTTPリクエストやサーバーからのレスポンスに付加されるメタデータ（データに関する情報を含むデータ）である
    - 一般的な「リクエストヘッダー」には認証情報が含まれる
    - 一般的な「レスポンスヘッダー」には、次節で示すように、サーバーが返すデータのタイプ型が含まれる
- URLパラメーターはURLの末尾にあるキーと値のペアで、URLから何を返してほしいかを指定する

### 11.1.2 SDGs APIからのデータの取得

PythonでAPIを呼び出す最も簡単な方法は`Requests`ライブラリを用いることだ。

```python
import requests


response = requests.get("https://unstats.un.org/SDGAPI/v1/sdg/Goal/List")
```

```python
response.status_code
```

```python
response.headers["content-type"]
```

```python
response.json()
```
